# Extração de Textos de Imagens de Jornais

Este notebook extrai automaticamente os textos de imagens de páginas de jornal, usando:
- **OCR** (pytesseract) para uma leitura inicial do texto
- **API de visão** (OpenRouter + Modelo de escolha) para, a partir do OCR, extrair corretamente cada texto e estruturar os resultados

**Resultado:** um arquivo Excel com os artigos extraídos, contendo título, autor e texto completo de cada matéria.

---

### Formato esperado das imagens
Os arquivos de imagem devem seguir o padrão: `YYYY-MM-DD_PAGINA.jpg`  
Exemplo: `2012-06-04_002.jpg` (edição de 4 de junho de 2012, página 2)

## 2. Configuração

Preencha as informações abaixo antes de executar:

```python
from openai import OpenAI
from extractor import process_folder

# ──────────────────────────────────────────────────────────────────
API_KEY      = "sua-chave-aqui"          # Chave do OpenRouter
INPUT_FOLDER = "caminho/para/as/imagens" # Pasta com os JPGs
OUTPUT_XLSX  = "resultados.xlsx"         # Arquivo de saída
NEWSPAPER_NAME  = "FSP"                     # Nome do jornal
# ──────────────────────────────────────────────────────────────────

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=API_KEY,
)
```
